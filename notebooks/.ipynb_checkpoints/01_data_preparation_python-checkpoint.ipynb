# End-to-End E-Commerce Data Analysis (SQL + Python)

## Project Overview
This project performs an end-to-end analysis of an e-commerce dataset using **Python for data preparation** and **MySQL for analytical querying**.

The original dataset was provided in **Excel format**. Python was primarily used to:
- Load and inspect the raw datasets
- Perform basic structure validation
- Convert Excel files into CSV format suitable for MySQL ingestion

## Why Python + SQL?
- **Python** is used for initial data handling, format conversion, and validation.
- **MySQL** is used for scalable querying, aggregations, window functions, and business analysis.

Although Python can load data directly into MySQL, this approach becomes time-consuming for large datasets.
Therefore, an optimized method using **MySQL `LOAD DATA INFILE`** was chosen for faster ingestion.
This notebook still demonstrates the Python-based approach for completeness and learning purposes.

## Dataset Summary
The dataset includes the following tables:
- Customers
- Orders
- Order Items
- Payments
- Products
- Sellers
- Geolocation

This notebook focuses on **data preparation and conversion**, while advanced analytics are performed in SQL.
