# End-to-End E-Commerce Data Analysis (SQL + Python)

## Project Overview
This project performs an end-to-end analysis of an e-commerce dataset using **Python for data preparation** and **MySQL for analytical querying**.

The original dataset was provided in **Excel format**. Python was primarily used to:

## Loading Data into MySQL using Python (Demonstration)

Python can be used to load CSV data into MySQL using libraries such as `mysql-connector-python` or `SQLAlchemy`.

However, for large datasets, this row-by-row insertion approach is significantly slower compared to MySQL’s native bulk ingestion.

The following section demonstrates the Python-based method for learning purposes, while the actual project ingestion was performed using `LOAD DATA INFILE` in MySQL Workbench.


```python
"""
DEMONSTRATION ONLY – NOT EXECUTED

This example shows how Python could load CSV data into MySQL.
The project uses MySQL LOAD DATA INFILE instead for performance reasons.
"""

import mysql.connector

connection = mysql.connector.connect(
    host="localhost",
    user="your_username",
    password="your_password",
    database="ecommerce_db"
)

cursor = connection.cursor()

load_query = """
LOAD DATA LOCAL INFILE 'customers.csv'
INTO TABLE customers
FIELDS TERMINATED BY ','
ENCLOSED BY '"'
LINES TERMINATED BY '\n'
IGNORE 1 ROWS;
"""

cursor.execute(load_query)
connection.commit()

cursor.close()
connection.close()
```

### Final Ingestion Approach Used in This Project

Although Python-based ingestion is demonstrated above, the actual data loading was performed using **MySQL LOAD DATA INFILE** directly within MySQL Workbench.

This decision was made to:
- Reduce ingestion time for large datasets
- Improve reliability
- Follow industry-standard data engineering practices
